```python
def energy(x, T):

    energy = torch.einsum('ijk,bi,bj,bk->b', T, x, x, x)

    return energy / 6
```

```python
min_e_list = []

greedy_min_list = []
greedy_median_list = []

polySBM_min_list = []
polySBM_median_list = []

polysimCIM_min_list = []
polysimCIM_median_list = []

polyVelox_min_list = []
polyVelox_median_list = []

QUBO_SBM_min_list = []
QUBO_SBM_median_list = []

Velox_q_min_list = []
Velox_q_median_list = []

for j in range(20):
    print(j)
    
    greedy_min = []
    greedy_median = []

    polySBM_min = []
    polySBM_median = []

    polysimCIM_min = []
    polysimCIM_median = []

    polyVelox_min = []
    polyVelox_median = []

    QUBO_SBM_min = []
    QUBO_SBM_median = []

    Velox_q_min = []
    Velox_q_median = []
    
    for i in range(49):
        T = np.load(f"J_T_h/T_{j}.npy", allow_pickle=True)
        
        
        
        data = np.load(f"solution_data/greedy_{j}.npy", allow_pickle=True)    
        e_vec = energy(torch.tensor(data[i], dtype=torch.float32, device="cuda"), torch.tensor(T, dtype=torch.float32, device="cuda"))
        greedy_min.append(float(torch.min(e_vec)))
        greedy_median.append(float(torch.median(e_vec)))


        data = np.load(f"solution_data/polySBM_{j}.npy", allow_pickle=True)
        e_vec = energy(torch.tensor(data[i], dtype=torch.float32, device="cuda"), torch.tensor(T, dtype=torch.float32, device="cuda"))
        polySBM_min.append(float(torch.min(e_vec)))
        polySBM_median.append(float(torch.median(e_vec)))


        data = np.load(f"solution_data/polysimCIM_{j}.npy", allow_pickle=True)
        e_vec = energy(torch.tensor(data[i], dtype=torch.float32, device="cuda"), torch.tensor(T, dtype=torch.float32, device="cuda"))
        polysimCIM_min.append(float(torch.min(e_vec)))
        polysimCIM_median.append(float(torch.median(e_vec)))


        data = np.load(f"solution_data/polyVelox_{j}.npy", allow_pickle=True)
        e_vec = energy(torch.tensor(data[i], dtype=torch.float32, device="cuda"), torch.tensor(T, dtype=torch.float32, device="cuda"))
        polyVelox_min.append(float(torch.min(e_vec)))
        polyVelox_median.append(float(torch.median(e_vec)))

        data = np.load(f"solution_data/QUBO_SBM_{j}.npy", allow_pickle=True)
        e_vec = energy(torch.tensor(data[i], dtype=torch.float32, device="cuda"), torch.tensor(T, dtype=torch.float32, device="cuda"))
        QUBO_SBM_min.append(float(torch.min(e_vec)))
        QUBO_SBM_median.append(float(torch.median(e_vec)))

        data = np.load(f"solution_data/Velox_q_{j}.npy", allow_pickle=True)
        e_vec = energy(torch.tensor(data[i], dtype=torch.float32, device="cuda"), torch.tensor(T, dtype=torch.float32, device="cuda"))
        Velox_q_min.append(float(torch.min(e_vec)))
        Velox_q_median.append(float(torch.median(e_vec)))
        
    
    best_e = min(np.min(Velox_q_min), np.min(QUBO_SBM_min), np.min(polyVelox_min), np.min(polysimCIM_min), np.min(polySBM_min), np.min(greedy_min))
    
    min_e_list.append(best_e)
    
    greedy_min_list.append(np.array(greedy_min) / best_e)
    greedy_median_list.append(np.array(greedy_median) / best_e)

    polySBM_min_list.append(np.array(polySBM_min) / best_e)
    polySBM_median_list.append(np.array(polySBM_median) / best_e)

    polysimCIM_min_list.append(np.array(polysimCIM_min) / best_e)
    polysimCIM_median_list.append(np.array(polysimCIM_median) / best_e)

    polyVelox_min_list.append(np.array(polyVelox_min) / best_e)
    polyVelox_median_list.append(np.array(polyVelox_median) / best_e)

    QUBO_SBM_min_list.append(np.array(QUBO_SBM_min) / best_e)
    QUBO_SBM_median_list.append(np.array(QUBO_SBM_median) / best_e)

    Velox_q_min_list.append(np.array(Velox_q_min) / best_e)
    Velox_q_median_list.append(np.array(Velox_q_median) / best_e)
    
```
