```python
# Per-run solver output for the three instance families, sizes 10-100.
import os
DATA = os.environ.get('PUBO_RUNS',
                      os.path.join('..', '..', 'data', 'benchmark_runs'))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import seaborn as sns
import scipy.stats as sts
font = {'family' : 'normal',
        'weight' : 'bold',
        'size'   : 14}

matplotlib.rc('font', **font)
```

```python
colors_ar = ['blue', 'red', 'brown', 'orange','cyan', 'green']
algos_ar = ['PolySimCIM','Hopfield',  'GD',  'Greedy', 'Leleu', 'TGD+CC']   
markers_ar = ['o', 'o','o', 'o', 'o', 'o']
linestyle_ar = ['-',  'dotted', '--',  (0, (5,10)), '-.', '--']
n_algos = np.shape(algos_ar)[0]

maxs=10
optimum=np.zeros((n_algos, 10,10))
opt_histogram=np.zeros((n_algos,10,10,2000))
final_opt=np.zeros((n_algos,10))
max_quantile_25 = np.zeros((n_algos,10, 10))
max_quantile_75 = np.zeros((n_algos,10, 10))
median_quantile_25 = np.zeros((n_algos,10,10))
median_quantile_75 = np.zeros((n_algos,10,10))
median = np.zeros((n_algos, 10,10))
quantile_25 = np.zeros((n_algos, 10,10))
quantile_75 = np.zeros((n_algos, 10,10))
final_median=np.zeros((n_algos,10))
sizes=np.arange(10, 110, 10)
num_array = [0, 1, 3, 4, 5]
all_algos = np.zeros((n_algos, 2000))

for p in range(1, maxs+1):
    N = 10*p
    for l in range(10):
        all_algos[0]=np.load(DATA+'/sk/polysimcim/'+str(N)+'_'+str(l)+'.npy')
        all_algos[1]=np.load(DATA+'/sk/hopfield/'+str(N)+'_'+str(l)+'.npy')
        all_algos[2]=np.load(DATA+'/sk/polysimcim/'+str(N)+'_'+str(l)+'.npy')
        all_algos[3]=np.load(DATA+'/sk/greedy/'+str(N)+'_'+str(l)+'.npy')
        all_algos[4]=np.load(DATA+'/sk/leleu/'+str(N)+'_'+str(l)+'.npy')
        all_algos[5]=np.load(DATA+'/sk/tgd/'+str(N)+'_'+str(l)+'.npy')
        
        Norm=np.min([all_algos[0].min(), all_algos[1].min(), all_algos[2].min(), all_algos[3].min(), all_algos[4].min(), all_algos[5].min()])
        
        for i in range(6):
            optimum[i,p-1,l]=all_algos[i].min()/Norm
            opt_histogram[i,p-1,l]=sorted(all_algos[i]/Norm)
            median[i,p-1,l]=np.median(all_algos[i]/Norm)
            
        for j in range(6):
            final_opt[j,p-1]=np.median(optimum[j,p-1])
            final_median[j,p-1]=np.median(median[j,p-1])
            max_quantile_25[j,p-1,l] = np.quantile(optimum[j,p-1,l], 0.25)
            max_quantile_75[j,p-1,l] = np.quantile(optimum[j,p-1,l], 0.75)
            median_quantile_25[j,p-1,l] = np.quantile(median[j,p-1,l], 0.25)
            median_quantile_75[j,p-1,l] = np.quantile(median[j,p-1,l], 0.75)

fig=plt.figure(figsize=(60,100))

for l in range(10):
    plt.subplot(10,3,3*l+1)
    plt.grid(True)
    plt.title('Dense tensors, $J_{i_{1}i_{2}i_{3}}=\pm 1$\n(Class I)', fontweight="bold", fontsize=16)
    plt.title('Instance = '+ str(l), fontweight="bold", fontsize=16, loc='left')
    for i in num_array:
        plt.plot(sizes[:maxs], optimum[i,:maxs, l], marker = markers_ar[i], markersize=10, color =  colors_ar[i], linestyle=linestyle_ar[i], linewidth=3, label=algos_ar[i])
        plt.xticks(fontsize=16)
        plt.yticks(fontsize=16)

maxs=10
optimum=np.zeros((n_algos, 10,10))
opt_histogram=np.zeros((n_algos,10,10,2000))
final_opt=np.zeros((n_algos,10))
median=np.zeros((n_algos, 10,10))
final_median=np.zeros((n_algos,10))
sizes=np.arange(10, 110, 10)
for p in range(1, maxs+1):
    N = 10*p
    for l in range(10):
        all_algos[0]=np.load(DATA+'/dense/polysimcim/'+str(N)+'_'+str(l)+'.npy')
        all_algos[1]=np.load(DATA+'/dense/hopfield/'+str(N)+'_'+str(l)+'.npy')
        all_algos[2]=np.load(DATA+'/dense/polysimcim/'+str(N)+'_'+str(l)+'.npy')
        all_algos[3]=np.load(DATA+'/dense/greedy/'+str(N)+'_'+str(l)+'.npy')
        all_algos[4]=np.load(DATA+'/dense/leleu/'+str(N)+'_'+str(l)+'.npy')
        all_algos[5]=np.load(DATA+'/dense/tgd/'+str(N)+'_'+str(l)+'.npy')
        
        Norm=np.min([all_algos[0].min(), all_algos[1].min(), all_algos[2].min(), all_algos[3].min(), all_algos[4].min(), all_algos[5].min()])
        
        for i in range(6):
            optimum[i,p-1,l] = all_algos[i].min()/Norm
            opt_histogram[i,p-1,l]=sorted(all_algos[i]/Norm)
            median[i,p-1,l] = np.median(all_algos[i]/Norm)
            quantile_25[i, p-1, l] = np.quantile(all_algos[i]/Norm, 0.75)
            quantile_75[i, p-1, l] = np.quantile(all_algos[i]/Norm, 1.0)
            
    for j in range(6):
        final_opt[j,p-1]=np.median(optimum[j,p-1])
        final_median[j,p-1]=np.median(median[j,p-1])
        max_quantile_25[j,p-1] = np.quantile(optimum[j,p-1], 0.25)
        max_quantile_75[j,p-1] = np.quantile(optimum[j,p-1], 0.75)
        median_quantile_25[j,p-1] = np.quantile(median[j,p-1], 0.25)
        median_quantile_75[j,p-1] = np.quantile(median[j,p-1], 0.75)

for l in range(10):
    plt.subplot(10,3,3*l+2)
    plt.grid(True)
    plt.title('Dense random tensors\n(Class II)', fontweight="bold", fontsize=16)
    plt.title('Instance = '+ str(l), fontweight="bold", fontsize=16, loc='left')
    for i in num_array:
        plt.plot(sizes[:maxs], optimum[i,:maxs, l], marker = markers_ar[i], markersize=10, color =  colors_ar[i], linestyle=linestyle_ar[i], linewidth=3, label=algos_ar[i])
        plt.xticks(fontsize=16)
        plt.yticks(fontsize=16)

for p in range(1, maxs+1):
    N = 10*p
    for l in range(10):
        all_algos[0]=np.load(DATA+'/sparse/polysimcim/'+str(N)+'_'+str(l)+'.npy')
        all_algos[1]=np.load(DATA+'/sparse/hopfield/'+str(N)+'_'+str(l)+'.npy')
        all_algos[2]=np.load(DATA+'/sparse/polysimcim/'+str(N)+'_'+str(l)+'.npy')
        all_algos[3]=np.load(DATA+'/sparse/greedy/'+str(N)+'_'+str(l)+'.npy')
        all_algos[4]=np.load(DATA+'/sparse/leleu/'+str(N)+'_'+str(l)+'.npy')
        all_algos[5]=np.load(DATA+'/sparse/tgd/'+str(N)+'_'+str(l)+'.npy')
        
        Norm=np.min([all_algos[0].min(), all_algos[1].min(), all_algos[2].min(), all_algos[3].min(), all_algos[4].min(), all_algos[5].min()])
        
        for i in range(6):
            optimum[i,p-1,l]=all_algos[i].min()/Norm
            opt_histogram[i,p-1,l]=sorted(all_algos[i]/Norm)
            median[i,p-1,l] = np.median(all_algos[i]/Norm)
            quantile_25[i, p-1, l] = np.quantile(all_algos[i]/Norm, 0.25)
            quantile_75[i, p-1, l] = np.quantile(all_algos[i]/Norm, 0.75)
            
    for j in range(6):
        final_opt[j,p-1]=np.median(optimum[j,p-1])
        final_median[j,p-1]=np.median(median[j,p-1])
        max_quantile_25[j,p-1] = np.quantile(optimum[j,p-1], 0.25)
        max_quantile_75[j,p-1] = np.quantile(optimum[j,p-1], 0.75)
        median_quantile_25[j,p-1] = np.quantile(median[j,p-1], 0.25)
        median_quantile_75[j,p-1] = np.quantile(median[j,p-1], 0.75)
        
for l in range(10):
    plt.subplot(10,3,3*l+3)
    plt.grid(True)
    plt.title('Sparse random tensors\n(Class III)', fontweight="bold", fontsize=16)
    plt.title('Instance = '+ str(l), fontweight="bold", fontsize=16, loc='left')
    for i in num_array:
        plt.plot(sizes[:maxs], optimum[i,:maxs, l], marker = markers_ar[i], markersize=10, color =  colors_ar[i], linestyle=linestyle_ar[i], linewidth=3, label=algos_ar[i])
        plt.xticks(fontsize=16)
        plt.yticks(fontsize=16)

plt.savefig('../../figures/benchmark_per_instance.pdf', bbox_inches='tight')
```
